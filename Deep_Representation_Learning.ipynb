```python
# ===============================================================
# 🌾 Project Samarth – Stable DCGAN for Animals Dataset (Research Quality)
# ===============================================================

import os, torch, torch.nn as nn, torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
from torch.utils.data import DataLoader
from tqdm import tqdm
import matplotlib.pyplot as plt, numpy as np

# ------------------ PATHS ------------------
data_dir = "/content/drive/MyDrive/TASK 1/animals"
project_path = "/content/drive/MyDrive/TASK 1/DCGAN_Final"
os.makedirs(project_path, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Running on:", device)

# ------------------ DATASET ------------------
transform = transforms.Compose([
    transforms.Resize((64, 64)),               # Smaller = faster + stable
    transforms.RandomHorizontalFlip(p=0.5),    # Minor augmentation for diversity
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

dataset = datasets.ImageFolder(root=data_dir, transform=transform)
dataloader = DataLoader(dataset, batch_size=128, shuffle=True, num_workers=2, pin_memory=True)
print(f"✅ Loaded {len(dataset)} images across {len(dataset.classes)} classes.")

# ------------------ MODEL PARAMS ------------------
nz, ngf, ndf = 100, 64, 64

# ------------------ GENERATOR ------------------
class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.main = nn.Sequential(
            nn.ConvTranspose2d(nz, ngf*8, 4, 1, 0, bias=False),
            nn.BatchNorm2d(ngf*8),
            nn.ReLU(True),

            nn.ConvTranspose2d(ngf*8, ngf*4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*4),
            nn.ReLU(True),

            nn.ConvTranspose2d(ngf*4, ngf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*2),
            nn.ReLU(True),

            nn.ConvTranspose2d(ngf*2, ngf, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf),
            nn.ReLU(True),

            nn.ConvTranspose2d(ngf, 3, 4, 2, 1, bias=False),
            nn.Tanh()
        )
    def forward(self, z): return self.main(z)

# ------------------ DISCRIMINATOR ------------------
class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()
        self.main = nn.Sequential(
            nn.utils.spectral_norm(nn.Conv2d(3, ndf, 4, 2, 1, bias=False)),
            nn.LeakyReLU(0.2, inplace=True),

            nn.utils.spectral_norm(nn.Conv2d(ndf, ndf*2, 4, 2, 1, bias=False)),
            nn.BatchNorm2d(ndf*2),
            nn.LeakyReLU(0.2, inplace=True),

            nn.utils.spectral_norm(nn.Conv2d(ndf*2, ndf*4, 4, 2, 1, bias=False)),
            nn.BatchNorm2d(ndf*4),
            nn.LeakyReLU(0.2, inplace=True),

            nn.utils.spectral_norm(nn.Conv2d(ndf*4, ndf*8, 4, 2, 1, bias=False)),
            nn.BatchNorm2d(ndf*8),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(ndf*8, 1, 4, 1, 0, bias=False),
            nn.AdaptiveAvgPool2d(1),
            nn.Sigmoid()
        )
    def forward(self, x): return self.main(x).view(-1)

# ------------------ INITIALIZATION ------------------
def weights_init(m):
    cname = m.__class__.__name__
    if cname.find('Conv')!=-1: nn.init.normal_(m.weight.data,0.0,0.02)
    elif cname.find('BatchNorm')!=-1:
        nn.init.normal_(m.weight.data,1.0,0.02); nn.init.constant_(m.bias.data,0)

netG, netD = Generator().to(device), Discriminator().to(device)
netG.apply(weights_init); netD.apply(weights_init)

# ------------------ HYPERPARAMS ------------------
criterion = nn.BCELoss()
optimizerD = optim.Adam(netD.parameters(), lr=0.0002, betas=(0.5,0.999))
optimizerG = optim.Adam(netG.parameters(), lr=0.0001, betas=(0.5,0.999))
epochs = 50
fixed_noise = torch.randn(100, nz, 1, 1, device=device)

# ------------------ TRAINING ------------------
print("🚀 Final Stable DCGAN Training Started...")
G_losses, D_losses = [], []

for epoch in range(epochs):
    for i,(real,_) in enumerate(tqdm(dataloader,leave=False)):
        real = real.to(device)
        b = real.size(0)
        real_label = torch.full((b,), 0.9, device=device)  # label smoothing
        fake_label = torch.zeros(b, device=device)

        # ---- Train D ----
        netD.zero_grad()
        out_real = netD(real)
        lossD_real = criterion(out_real, real_label)

        z = torch.randn(b, nz, 1, 1, device=device)
        fake = netG(z)
        out_fake = netD(fake.detach())
        lossD_fake = criterion(out_fake, fake_label)
        lossD = lossD_real + lossD_fake
        lossD.backward()
        torch.nn.utils.clip_grad_norm_(netD.parameters(), 1.0)
        optimizerD.step()

        # ---- Train G ----
        netG.zero_grad()
        out_fake = netD(fake)
        lossG = criterion(out_fake, real_label)
        lossG.backward()
        torch.nn.utils.clip_grad_norm_(netG.parameters(), 1.0)
        optimizerG.step()

    G_losses.append(lossG.item()); D_losses.append(lossD.item())
    print(f"Epoch[{epoch+1}/{epochs}] | Loss_D={lossD.item():.4f} | Loss_G={lossG.item():.4f}")

    # ---- Save grid every 5 epochs ----
    if (epoch+1)%5==0:
        with torch.no_grad():
            fake_imgs = netG(fixed_noise).detach().cpu()
        grid = make_grid(fake_imgs[:100], nrow=10, normalize=True)
        save_image(grid, f"{project_path}/epoch_{epoch+1}_grid.png")

print("✅ Training Complete!")

# ------------------ VISUALIZATION ------------------
plt.figure(figsize=(10,5))
plt.plot(G_losses,label="G_Loss"); plt.plot(D_losses,label="D_Loss")
plt.legend(); plt.title("Stable DCGAN Loss Curve")
plt.savefig(f"{project_path}/loss_curve_final.png"); plt.show()

# ------------------ SAVE MODELS ------------------
torch.save(netG.state_dict(), f"{project_path}/generator_final.pth")
torch.save(netD.state_dict(), f"{project_path}/discriminator_final.pth")
print("✅ Models & results saved in:", project_path)

# ------------------ SAMPLE OUTPUT ------------------
with torch.no_grad():
    fake_imgs = netG(fixed_noise).detach().cpu()
grid = make_grid(fake_imgs[:100], nrow=10, normalize=True)
plt.figure(figsize=(10,10))
plt.imshow(np.transpose(grid,(1,2,0)))
plt.axis('off'); plt.title("10×10 Generated Animal Images Grid (Final)")
plt.show()
```

```python
# ===============================================================
# ✅ Latent Space Interpolation — Linear & Non-Linear (Spherical)
# ===============================================================

import torch
import matplotlib.pyplot as plt
import numpy as np
from torchvision.utils import make_grid, save_image
import os

# Use same latent dimension as your trained generator
# Ensure netG and nz are defined from your training notebook
project_path = "/content/drive/MyDrive/TASK 1/DCGAN_Final"

# Generate two random latent vectors
z1 = torch.randn(1, nz, 1, 1, device=device)
z2 = torch.randn(1, nz, 1, 1, device=device)

# Define interpolation range (10 steps)
alphas = torch.linspace(0, 1, 10)

# ------------------ 1️⃣ Linear Interpolation ------------------
linear_interpolations = []

with torch.no_grad():
    for alpha in alphas:
        # Move in a straight line between z1 and z2
        z = (1 - alpha) * z1 + alpha * z2
        fake_img = netG(z).cpu()
        linear_interpolations.append(fake_img)

# Combine into a single grid
grid_linear = make_grid(torch.cat(linear_interpolations), nrow=10, normalize=True)

# Display
plt.figure(figsize=(12,4))
plt.imshow(np.transpose(grid_linear, (1,2,0)))
plt.axis('off')
plt.title("1️⃣ Linear Latent Space Interpolation (z₁ → z₂)")
plt.show()

# Save image
save_image(grid_linear, os.path.join(project_path, "latent_interpolation_linear.png"))
print("✅ Saved: latent_interpolation_linear.png")

# ------------------ 2️⃣ Non-Linear (Spherical) Interpolation ------------------

def slerp(val, low, high):
    """Spherical linear interpolation (for smooth curve on hypersphere)"""
    low_flat = low.flatten()
    high_flat = high.flatten()
    omega = torch.acos(torch.dot(low_flat, high_flat) / (torch.norm(low_flat) * torch.norm(high_flat)))
    so = torch.sin(omega)
    if so == 0:
        return (1.0 - val) * low + val * high  # fallback to linear if identical
    return (torch.sin((1.0 - val) * omega) / so) * low + (torch.sin(val * omega) / so) * high

alphas = torch.linspace(0, 1, 10)
spherical_interpolations = []

with torch.no_grad():
    for alpha in alphas:
        # Move along curved path (constant radius)
        z = slerp(alpha, z1.squeeze(), z2.squeeze()).unsqueeze(0).unsqueeze(2).unsqueeze(3)
        fake_img = netG(z).cpu()
        spherical_interpolations.append(fake_img)

# Combine into grid
grid_spherical = make_grid(torch.cat(spherical_interpolations), nrow=10, normalize=True)

# Display
plt.figure(figsize=(12,4))
plt.imshow(np.transpose(grid_spherical, (1,2,0)))
plt.axis('off')
plt.title("2️⃣ Non-Linear (Spherical) Latent Space Interpolation")
plt.show()

# Save image
save_image(grid_spherical, os.path.join(project_path, "latent_interpolation_spherical.png"))
print("✅ Saved: latent_interpolation_spherical.png")
```

```python
# ===============================================================
# ✅ TASK 6 — CONDITIONAL GAN (FINAL RESEARCH-GRADE VERSION)
# ===============================================================

import os, random, torch, torch.nn as nn, torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms

# ---------------------------------------------------------------
# 1️⃣  Setup
# ---------------------------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Using device: {device}")

data_dir = "/content/drive/MyDrive/TASK 1/animals"
transform = transforms.Compose([
    transforms.Resize((64,64)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
])
dataset = datasets.ImageFolder(root=data_dir, transform=transform)
print(f"✅ Dataset loaded: {len(dataset)} images across {len(dataset.classes)} classes.")

# ---------------------------------------------------------------
# 2️⃣  Select 20 random classes + remap labels
# ---------------------------------------------------------------
random.seed(42)
all_classes = dataset.classes
selected_classes = random.sample(all_classes, 20)
class_to_new = {cls: i for i, cls in enumerate(selected_classes)}
print("🎯 Selected 20 classes:", selected_classes)
print("✅ Class Reindex Mapping:", class_to_new)

indices = [i for i, (_, label) in enumerate(dataset.samples)
           if dataset.classes[label] in selected_classes]
subset_20 = Subset(dataset, indices)

class SubsetRemap(torch.utils.data.Dataset):
    def __init__(self, subset, dataset, mapping):
        self.subset, self.dataset, self.mapping = subset, dataset, mapping
    def __getitem__(self, idx):
        img, label = self.subset[idx]
        clsname = self.dataset.classes[label]
        return img, self.mapping[clsname]
    def __len__(self): return len(self.subset)

subset_mapped = SubsetRemap(subset_20, dataset, class_to_new)
subset_loader = DataLoader(subset_mapped, batch_size=64, shuffle=True, num_workers=2, pin_memory=True)

# ---------------------------------------------------------------
# 3️⃣  Models
# ---------------------------------------------------------------
num_classes, nz, ngf, ndf = 20, 100, 64, 64

# ---------- Generator ----------
class Generator_cGAN(nn.Module):
    def __init__(self):
        super().__init__()
        self.label_emb = nn.Embedding(num_classes, nz)
        self.main = nn.Sequential(
            nn.ConvTranspose2d(nz*2, ngf*8, 4, 1, 0, bias=False),
            nn.BatchNorm2d(ngf*8), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*8, ngf*4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*4), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*4, ngf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*2), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*2, ngf, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf), nn.ReLU(True),
            nn.ConvTranspose2d(ngf, 3, 4, 2, 1, bias=False),
            nn.Tanh()
        )
    def forward(self, z, labels):
        label_vec = self.label_emb(labels)
        gen_input = torch.cat((z.squeeze(), label_vec), -1).unsqueeze(2).unsqueeze(3)
        return self.main(gen_input)

# ---------- Discriminator (Fixed Output Shape) ----------
class Discriminator_cGAN(nn.Module):
    def __init__(self):
        super().__init__()
        self.label_emb = nn.Embedding(num_classes, 64*64)
        self.main = nn.Sequential(
            nn.Conv2d(4, ndf, 4, 2, 1, bias=False),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf, ndf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf*2), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf*2, ndf*4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf*4), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf*4, 1, 4, 1, 0, bias=False),
            nn.Sigmoid(),
            nn.AdaptiveAvgPool2d(1)   # ✅ ensures 1 output per image
        )
    def forward(self, img, labels):
        label_map = self.label_emb(labels).view(labels.size(0), 1, 64, 64)
        x = torch.cat((img, label_map), 1)
        return self.main(x).view(-1)  # shape [batch]

# ---------------------------------------------------------------
# 4️⃣  Initialize + Train
# ---------------------------------------------------------------
torch.cuda.empty_cache()
netG_c, netD_c = Generator_cGAN().to(device), Discriminator_cGAN().to(device)
criterion = nn.BCELoss()
optimizerD = optim.Adam(netD_c.parameters(), lr=0.0002, betas=(0.5,0.999))
optimizerG = optim.Adam(netG_c.parameters(), lr=0.0001, betas=(0.5,0.999))
print("✅ Conditional GAN initialized successfully on GPU.")

epochs = 10
for epoch in range(epochs):
    loss_D_epoch, loss_G_epoch = 0, 0
    for imgs, labels in subset_loader:
        imgs, labels = imgs.to(device), labels.to(device, dtype=torch.long)
        b_size = imgs.size(0)
        real_label = torch.full((b_size,), 0.9, device=device)
        fake_label = torch.zeros(b_size, device=device)

        # ----- Train Discriminator -----
        netD_c.zero_grad()
        out_real = netD_c(imgs, labels)
        lossD_real = criterion(out_real, real_label)

        z = torch.randn(b_size, nz, 1, 1, device=device)
        fake = netG_c(z, labels)
        out_fake = netD_c(fake.detach(), labels)
        lossD_fake = criterion(out_fake, fake_label)

        lossD = lossD_real + lossD_fake
        lossD.backward()
        optimizerD.step()

        # ----- Train Generator -----
        netG_c.zero_grad()
        out_fake = netD_c(fake, labels)
        lossG = criterion(out_fake, real_label)
        lossG.backward()
        optimizerG.step()

        loss_D_epoch += lossD.item()
        loss_G_epoch += lossG.item()

    print(f"Epoch [{epoch+1}/{epochs}] "
          f"| Loss_D: {loss_D_epoch/len(subset_loader):.3f} "
          f"| Loss_G: {loss_G_epoch/len(subset_loader):.3f}")

print("✅ Task 6 complete — Conditional GAN trained successfully (Stable Version).")
```

```python
# ===============================================================
# ✅ TASK 7–9 — CLASSIFICATION + AUGMENTATION (FINAL VERSION)
# ===============================================================
import torch, torch.nn as nn, torch.optim as optim
from torchvision import models, datasets, transforms
from torch.utils.data import DataLoader, Subset, ConcatDataset
from sklearn.metrics import accuracy_score, f1_score
import numpy as np
from tqdm import tqdm

# ---------------------------------------------------------------
# 🔹 Common transforms
# ---------------------------------------------------------------
train_tfms = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
])
test_tfms = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
])

data_dir = "/content/drive/MyDrive/TASK 1/animals"
full_dataset = datasets.ImageFolder(root=data_dir, transform=train_tfms)
print(f"✅ Full dataset: {len(full_dataset)} images, {len(full_dataset.classes)} classes.")

# ---------------------------------------------------------------
# ⚙️ Helper: train / evaluate loops
# ---------------------------------------------------------------
def train_classifier(model, loader, criterion, optimizer, epochs=5):
    model.train()
    for ep in range(epochs):
        total, correct, running_loss = 0, 0, 0
        for imgs, labels in tqdm(loader, desc=f"Epoch {ep+1}/{epochs}"):
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            out = model(imgs)
            loss = criterion(out, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()*imgs.size(0)
            preds = out.argmax(1)
            correct += (preds==labels).sum().item()
            total += labels.size(0)
        print(f"Loss: {running_loss/total:.4f} | Acc: {100*correct/total:.2f}%")

def evaluate(model, loader):
    model.eval(); preds_all, labels_all = [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs = imgs.to(device)
            out = model(imgs)
            preds_all.extend(out.argmax(1).cpu().numpy())
            labels_all.extend(labels.numpy())
    acc = accuracy_score(labels_all, preds_all)
    f1  = f1_score(labels_all, preds_all, average="macro")
    return acc, f1

# ===============================================================
# 🧠 TASK 7 — Fine-tune ResNet on 90 classes
# ===============================================================
train_loader = DataLoader(full_dataset, batch_size=64, shuffle=True, num_workers=2)

resnet_base = models.resnet50(weights="IMAGENET1K_V1")
resnet_base.fc = nn.Linear(resnet_base.fc.in_features, len(full_dataset.classes))
resnet_base = resnet_base.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet_base.parameters(), lr=1e-4)

print("🚀 Training ResNet on 90 classes (Task 7)...")
train_classifier(resnet_base, train_loader, criterion, optimizer, epochs=5)
acc_90, f1_90 = evaluate(resnet_base, train_loader)
print(f"✅ Task 7 → Accuracy: {acc_90*100:.2f}% | F1: {f1_90:.3f}")

# ===============================================================
# 🧩 TASK 8 — Train on 20-class subset
# ===============================================================
indices_20 = [i for i, (_, lbl) in enumerate(full_dataset.samples)
              if full_dataset.classes[lbl] in selected_classes]
subset_20 = Subset(full_dataset, indices_20)
subset_loader = DataLoader(subset_20, batch_size=64, shuffle=True, num_workers=2)

resnet_20 = models.resnet50(weights="IMAGENET1K_V1")
resnet_20.fc = nn.Linear(resnet_20.fc.in_features, 20)
resnet_20 = resnet_20.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet_20.parameters(), lr=1e-4)

print("\n🚀 Training ResNet on 20-class subset (Task 8)...")
train_classifier(resnet_20, subset_loader, criterion, optimizer, epochs=5)
acc_20, f1_20 = evaluate(resnet_20, subset_loader)
print(f"✅ Task 8 → Accuracy: {acc_20*100:.2f}% | F1: {f1_20:.3f}")

# ===============================================================
# 🧬 TASK 9 — cGAN Augmentation + Retraining
# ===============================================================
from torchvision.utils import save_image
import tempfile, os

aug_dir = tempfile.mkdtemp(prefix="aug_data_")
os.makedirs(os.path.join(aug_dir,"train"), exist_ok=True)

print("\n🎨 Generating 100 synthetic images per class from cGAN...")
netG_c.eval()
with torch.no_grad():
    for cls_name, cls_idx in class_to_new.items():
        cls_path = os.path.join(aug_dir,"train",cls_name)
        os.makedirs(cls_path, exist_ok=True)
        for i in range(100):
            z = torch.randn(1, nz, 1, 1, device=device)
            lbl = torch.tensor([cls_idx], device=device)
            fake_img = netG_c(z, lbl)
            save_image(fake_img, os.path.join(cls_path, f"fake_{i}.png"), normalize=True)

print("✅ Synthetic augmentation completed.")

aug_dataset = datasets.ImageFolder(root=os.path.join(aug_dir,"train"), transform=train_tfms)
combined_dataset = ConcatDataset([subset_20, aug_dataset])
combined_loader = DataLoader(combined_dataset, batch_size=64, shuffle=True, num_workers=2)

resnet_aug = models.resnet50(weights="IMAGENET1K_V1")
resnet_aug.fc = nn.Linear(resnet_aug.fc.in_features, 20)
resnet_aug = resnet_aug.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet_aug.parameters(), lr=1e-4)

print("\n🚀 Retraining ResNet on augmented data (Task 9)...")
train_classifier(resnet_aug, combined_loader, criterion, optimizer, epochs=5)
acc_aug, f1_aug = evaluate(resnet_aug, combined_loader)
print(f"✅ Task 9 → Accuracy (after augmentation): {acc_aug*100:.2f}% | F1: {f1_aug:.3f}")

# ---------------------------------------------------------------
# 📊 Compare Results
# ---------------------------------------------------------------
print("\n📈 Performance Summary")
print(f"Task 7 (90-class): Acc {acc_90*100:.2f}% | F1 {f1_90:.3f}")
print(f"Task 8 (20-class): Acc {acc_20*100:.2f}% | F1 {f1_20:.3f}")
print(f"Task 9 (+ Augmentation): Acc {acc_aug*100:.2f}% | F1 {f1_aug:.3f}")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import datasets, transforms

data_dir = "/content/drive/MyDrive/TASK 1/animals"
transform = transforms.Compose([
    transforms.Resize((96,96)),   # smaller for CPU speed
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
])

dataset = datasets.ImageFolder(root=data_dir, transform=transform)
print(f"✅ Dataset loaded: {len(dataset)} images across {len(dataset.classes)} classes.")

selected_classes = ['swan', 'coyote', 'bear', 'hedgehog', 'gorilla', 'goat', 'deer', 'cow',
                    'wolf', 'reindeer', 'chimpanzee', 'sheep', 'orangutan', 'bee', 'fox',
                    'goldfish', 'pigeon', 'sparrow', 'sandpiper', 'flamingo']

indices_20 = [i for i, (_, lbl) in enumerate(dataset.samples)
              if dataset.classes[lbl] in selected_classes]
class_to_new = {cls: i for i, cls in enumerate(selected_classes)}

class SubsetRemap(Dataset):
    def __init__(self, subset, dataset, mapping):
        self.subset, self.dataset, self.mapping = subset, dataset, mapping
    def __getitem__(self, idx):
        img, label = self.subset[idx]
        clsname = self.dataset.classes[label]
        return img, self.mapping[clsname]
    def __len__(self): return len(self.subset)

subset_20_raw = Subset(dataset, indices_20)
subset_20 = SubsetRemap(subset_20_raw, dataset, class_to_new)
subset_loader = DataLoader(subset_20, batch_size=16, shuffle=True, num_workers=2)

print(f"✅ 20-class subset rebuilt: {len(subset_20)} samples across {len(selected_classes)} classes.")
```

```python
from torchvision import models
import torch.nn as nn, torch.optim as optim
from tqdm import tqdm
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

device = torch.device("cpu")  # ensure CPU mode

# ✅ Define a smaller ResNet for CPU
resnet_20 = models.resnet18(weights="IMAGENET1K_V1")
resnet_20.fc = nn.Linear(resnet_20.fc.in_features, 20)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet_20.parameters(), lr=1e-4)

def train_classifier(model, loader, criterion, optimizer, epochs=3):
    model.train()
    for ep in range(epochs):
        total, correct, running_loss = 0, 0, 0
        for imgs, labels in tqdm(loader, desc=f"Epoch {ep+1}/{epochs}"):
            optimizer.zero_grad()
            out = model(imgs)
            loss = criterion(out, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()*imgs.size(0)
            preds = out.argmax(1)
            correct += (preds==labels).sum().item()
            total += labels.size(0)
        print(f"Loss: {running_loss/total:.4f} | Acc: {100*correct/total:.2f}%")

def evaluate(model, loader):
    model.eval(); preds_all, labels_all = [], []
    with torch.no_grad():
        for imgs, labels in loader:
            out = model(imgs)
            preds_all.extend(out.argmax(1).numpy())
            labels_all.extend(labels.numpy())
    acc = accuracy_score(labels_all, preds_all)
    f1  = f1_score(labels_all, preds_all, average="macro")
    return acc, f1

# 🚀 Train and Evaluate
print("\n🚀 Training ResNet18 on 20-class subset (Task 8, CPU mode)...")
train_classifier(resnet_20, subset_loader, criterion, optimizer, epochs=3)
acc_20, f1_20 = evaluate(resnet_20, subset_loader)
print(f"✅ Task 8 → Accuracy: {acc_20*100:.2f}% | F1: {f1_20:.3f}")
```
